#### Ejercicio 01:
Productos del más caro al más barato

```sql
SELECT
    nombre,
    precio
FROM productos
ORDER BY precio DESC;
```

#### Ejercicio 2:
Clientes de España o México cuyo nombre comienza con «M»

```sql
SELECT
    nombre,
    pais
FROM clientes
WHERE pais IN ('España', 'México')
  AND nombre LIKE 'M%';
```

#### Ejercicio 03:
Productos entre 20 y 60 € con stock **disponible**

```sql
SELECT
    nombre,
    precio,
    stock
FROM tienda.productos
WHERE precio BETWEEN 20 AND 60
  AND stock > 0;
```

#### Ejercicio 4: 
Clientes sin teléfono usando COALESCE

```sql
SELECT
    nombre,
    COALESCE(telefono, 'sin dato') AS telefono
FROM clientes
WHERE telefono IS NULL;
```

#### Ejercicio 5: 
Los 10 productos más caros con IVA

```sql
SELECT
    nombre,
    precio,
    ROUND(precio * 1.21, 2) AS precio_con_iva
FROM productos
ORDER BY precio DESC
LIMIT 10;
```

#### Ejercicio 6: 
Clasificar productos con CASE
- Menos de 20: barato.
- Entre 20 y 60: medio.
- Más de 60: caro.

```sql
SELECT
    nombre,
    precio,
    CASE
        WHEN precio < 20 THEN 'barato'
        WHEN precio <= 60 THEN 'medio'
        ELSE 'caro'
    END AS clasificacion
FROM productos
ORDER BY precio;
```

#### Ejercicio 7: 
Cantidad de países distintos

```sql
SELECT
    COUNT(DISTINCT pais) AS cantidad_paises
FROM tienda.clientes;
```
