```sql
CREATE SCHEMA IF NOT EXISTS tienda;
```

```sql
USE tienda;
```

```sql
SHOW TABLES;
```

```sql
CREATE TABLE IF NOT EXISTS categorias (
    id INT NOT NULL,
    nombre STRING NOT NULL
);

CREATE TABLE IF NOT EXISTS productos (
    id INT NOT NULL,
    nombre STRING NOT NULL,
    precio NUMERIC NOT NULL,
    stock INT,
    categoria_id INT NOT NULL
);

CREATE TABLE IF NOT EXISTS detalle_pedidos (
    pedido_id INT NOT NULL,
    producto_id INT NOT NULL,
    cantidad INT,
    precio_unit NUMERIC
);

CREATE TABLE IF NOT EXISTS pedidos (
    id INT NOT NULL,
    cliente_id INT NOT NULL,
    fecha DATE,
    total NUMERIC
);

CREATE TABLE IF NOT EXISTS clientes (
    id INT NOT NULL,
    nombre STRING NOT NULL,
    email STRING,
    pais STRING,
    telefono STRING
);
```

```sql
SELECT nombre, stock
FROM productos;
```

```sql
SELECT
    nombre,
    precio * stock AS precio_total_stock
FROM productos;
```

```sql
SELECT 
    nombre AS nombre_y_apellido,
    pais
FROM clientes;
```

```sql
SELECT * 
FROM productos
```

```sql
SELECT *
FROM productos
WHERE precio > 100;
```

```sql
SELECT *
FROM productos
WHERE precio <> 60; -- diferente
```

```sql
SELECT *
FROM productos
WHERE precio IN (25, 60, 16);
```

```sql
SELECT *
FROM productos
-- WHERE precio >= 10 AND precio <= 100;
WHERE precio BETWEEN 10 AND 100; -- ES LO MISMO A WHERE precio >= 10 AND precio <= 100;
```

```sql
SELECT nombre, precio 
FROM productos
WHERE (nombre = 'Vaqueros clásicos'
        OR nombre = 'Balón de fútbol')
    AND precio <= 20;
```

```sql
SELECT *
FROM clientes
WHERE NOT pais = 'ES';
```

```sql
SELECT *
FROM clientes
WHERE email LIKE '%@empresa%' -- el % indica que habrá contenido antes y después
```

```sql
SELECT *
FROM clientes
WHERE nombre LIKE '%ez' -- el % indica que habrá contenido antes pero no después
```

```sql
SELECT nombre, telefono
FROM clientes
WHERE telefono LIKE '11%';
```
