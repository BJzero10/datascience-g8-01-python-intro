```python
import pandas as pd
import matplotlib.pyplot as plt

datos_iris = pd.read_csv("../data/iris_dataset.csv")
datos_iris
```

```python
# 9. Histograma: longitud del sépalo

plt.figure(figsize=(7, 4))
plt.hist(
    datos_iris["sepal_length_cm"],
    bins=10,
    color="skyblue",
    edgecolor="black"
)

plt.title("Histograma de la longitud del sépalo")
plt.xlabel("Longitud del sépalo (cm)")
plt.ylabel("Frecuencia")
plt.grid(True, axis="y", linestyle="--", alpha=0.3)
plt.tight_layout()
plt.show()
```

Observación: La mayor parte de las observaciones se concentra en valores intermedios de la longitud del sépalo, aproximadamente entre 5.0 y 6.5 cm.

```python
# 10. Histograma y bins

plt.figure(figsize=(11, 4))

plt.subplot(1, 2, 1)
plt.hist(
    datos_iris["petal_length_cm"],
    bins=5,
    color="orange",
    edgecolor="black"
)
plt.title("Longitud del pétalo - 5 bins")
plt.xlabel("Longitud del pétalo (cm)")
plt.ylabel("Frecuencia")
plt.grid(True, axis="y", linestyle="--", alpha=0.3)

plt.subplot(1, 2, 2)
plt.hist(
    datos_iris["petal_length_cm"],
    bins=20,
    color="seagreen",
    edgecolor="black"
)
plt.title("Longitud del pétalo - 20 bins")
plt.xlabel("Longitud del pétalo (cm)")
plt.ylabel("Frecuencia")
plt.grid(True, axis="y", linestyle="--", alpha=0.3)

plt.tight_layout()
plt.show()
```

Observación: Con 5 bins los datos se agrupan en intervalos más grandes y la distribución se ve de forma general. Con 20 bins los intervalos son más pequeños y se aprecian más detalles.

```python
# 11. Boxplot por especie

datos_boxplot = [
    datos_iris[datos_iris["species"] == "setosa"]["sepal_length_cm"],
    datos_iris[datos_iris["species"] == "versicolor"]["sepal_length_cm"],
    datos_iris[datos_iris["species"] == "virginica"]["sepal_length_cm"]
]

plt.figure(figsize=(7, 4))
plt.boxplot(
    datos_boxplot,
    patch_artist=True,
    boxprops={"facecolor": "lightblue"}
)

plt.xticks([1, 2, 3], ["setosa", "versicolor", "virginica"])
plt.title("Longitud del sépalo por especie")
plt.xlabel("Especie")
plt.ylabel("Longitud del sépalo (cm)")
plt.grid(True, axis="y", linestyle="--", alpha=0.3)
plt.tight_layout()
plt.show()
```

Observación: Las medianas de la longitud del sépalo cambian entre especies. Setosa tiene la mediana más baja, mientras que virginica presenta una mediana mayor.

```python
# 12. Boxplot de variables numéricas

datos_boxplot = [
    datos_iris["sepal_length_cm"],
    datos_iris["sepal_width_cm"],
    datos_iris["petal_length_cm"],
    datos_iris["petal_width_cm"]
]

plt.figure(figsize=(8, 4))
plt.boxplot(
    datos_boxplot,
    patch_artist=True,
    boxprops={"facecolor": "lightgreen"}
)

plt.xticks([1, 2, 3, 4], ["Sépalo L.", "Sépalo A.", "Pétalo L.", "Pétalo A."])
plt.title("Boxplot de variables numéricas")
plt.xlabel("Variable")
plt.ylabel("Medición (cm)")
plt.grid(True, axis="y", linestyle="--", alpha=0.3)
plt.tight_layout()
plt.show()
```

Observación: Las variables tienen diferentes rangos de valores. Las longitudes suelen tener valores mayores que los anchos, por eso algunas cajas se ven más comprimidas.

```python
# 13. Subplots 1x2

plt.figure(figsize=(11, 4))

plt.subplot(1, 2, 1)
plt.scatter(
    datos_iris["sepal_length_cm"],
    datos_iris["sepal_width_cm"],
    color="steelblue",
    alpha=0.7
)
plt.title("Longitud vs. ancho del sépalo")
plt.xlabel("Longitud del sépalo (cm)")
plt.ylabel("Ancho del sépalo (cm)")
plt.grid(True, linestyle="--", alpha=0.3)

plt.subplot(1, 2, 2)
plt.hist(
    datos_iris["sepal_length_cm"],
    bins=10,
    color="orange",
    edgecolor="black"
)
plt.title("Histograma de longitud del sépalo")
plt.xlabel("Longitud del sépalo (cm)")
plt.ylabel("Frecuencia")
plt.grid(True, axis="y", linestyle="--", alpha=0.3)

plt.tight_layout()
plt.show()
```
