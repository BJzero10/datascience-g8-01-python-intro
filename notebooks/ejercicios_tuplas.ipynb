# Ejercicios de tuplas en Python

#### 1. Buscar el mayor y el menor valor
Dada la tupla numeros = (15, 8, 42, 3, 27, 19, 5), obtén el número mayor, el menor y la diferencia entre ambos.

```python
numeros = (15, 8, 42, 3, 27, 19, 5)
mayor = max(numeros)
menor = min(numeros)
print("El número mayor es:", mayor)
print("El número menor es:", menor)
```

#### 2. Contar ocurrencias
Dada la tupla datos = (5, 8, 5, 1, 9, 5, 2, 8, 5, 3), solicita un número al usuario y muestra cuántas veces aparece sin convertir la tupla en lista.

```python
datos = (5, 8, 5, 1, 9, 5, 2, 8, 5, 3)
num_usuario = int(input("Ingrese un número que desea contar: "))
conteo = datos.count(num_usuario)
print(f"El número {num_usuario} aparece {conteo} veces en la tupla.")
```

#### 3. Encontrar la primera posición
Dada la tupla nombres = ('Ana','Luis','Pedro','María','Luis','Carlos'), solicita un nombre y muestra su primera posición o un mensaje si no existe.

```python
nombres = ('Ana','Luis','Pedro','María','Luis','Carlos')
nombre_usuario = input("Ingrese un nombre que desea buscar: ")
if nombre_usuario in nombres:
    posicion = nombres.index(nombre_usuario)
    print(f"El nombre {nombre_usuario} se encuentra en la posición {posicion}.")
else:
    print(f"El nombre {nombre_usuario} no se encuentra en la tupla.")
```

#### 4. Estadísticas de una tupla
Con notas = (18,15,20,11,14,17,19), calcula promedio, máxima, mínima y cantidad de aprobados.

```python
notas = (18,15,20,11,14,17,19)
nota_max = max(notas)
nota_min = min(notas)
nota_promedio = sum(notas) / len(notas)
print("Nota máxima:", nota_max)
print("Nota mínima:", nota_min)
print("Nota promedio:", nota_promedio)
```

#### 5. Tuplas dentro de una lista
Con ventas=[('Lunes',120),('Martes',150),('Miércoles',95),('Jueves',180),('Viernes',210)], calcula el día con mayor venta, el total y el promedio.

```python
ventas=[('Lunes',120),('Martes',150),('Miércoles',95),('Jueves',180),('Viernes',210)]
dia_mayor_venta = max(ventas, key=lambda x: x[1])
total_ventas = sum(venta[1] for venta in ventas)
promedio_ventas = total_ventas / len(ventas)
print(f"El día con mayor venta es {dia_mayor_venta[0]}")
print(f"El total de ventas es: {total_ventas}")
print(f"El promedio de ventas es: {promedio_ventas}")
```
