```python
# convertir lista a array
lista = [1, 2, 3, 4, 5]
array = np.array(lista)
print(array)
print(type(array))
```

```python
lista_decimales = [1.5, 2.5, 3.5, 4.5, 5.5]
arr_decimales = np.array(lista_decimales)
print(arr_decimales)
print(type(arr_decimales))
print(arr_decimales.dtype)  # tipo de datos del array
```

```python
lista_enteros = [1, 2, 3, 4, 5]
arr_enteros = np.array(lista_enteros)
print(arr_enteros)
print(arr_enteros.dtype)  # tipo de datos del array
```

```python
lista_enteros = [1, 2, 3, 4, 5]
arr_enteros = np.array(lista_enteros, dtype=np.float64)  # especificar tipo de datos
print(arr_enteros)
print(arr_enteros.dtype)  # tipo de datos del array
```

```python
arr_float = arr_enteros.astype(np.float64)  # convertir a float64
print(arr_float)
print(arr_float.dtype)  # tipo de datos del array
```

```python
arr_str = arr_enteros.astype(str)  # convertir a string
print(arr_str)
print(arr_str.dtype)  # tipo de datos del array
```

### Dimensiones

```python
# ESCALAR
scalar = np.array(50)
print(scalar)
print(type(scalar))
print(scalar.ndim)  # número de dimensiones
```

```python
# VECTOR
vector = np.array([1, 2, 3, 4, 5])
print(vector)
print(type(vector))
print(vector.ndim)  # número de dimensiones
```

```python
# MATRIZ
matrix = np.array([[1, 2, 3], [4, 5, 6]])
print(matrix)
print(type(matrix))
print(matrix.ndim)  # número de dimensiones
```

### Generación de números aleatorios

```python
np.arange(0, 20, 2)  # rango de valores con paso de 2
```

```python
np.zeros(5, dtype="int64")  # array de ceros de tamaño 5 y tipo int64
```

```python
np.zeros((3, 4), dtype="float64")  # array de ceros de tamaño 3x4 y tipo float64
```

```python
np.ones(5, dtype="int64")  # array de unos de tamaño 5 y tipo int64
```

```python
np.ones((3, 4), dtype="float64")  # array de unos de tamaño 3x4 y tipo float64
```

```python
np.ones((3, 10, 2), dtype="float64")  # array de unos de tamaño 3x10x2 y tipo float64
```

```python
np.eye(3)  # matriz identidad de tamaño 3x3
```
