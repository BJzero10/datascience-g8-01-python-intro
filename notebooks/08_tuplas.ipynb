### TUPLAS

```python
lista = [1, 3, 5, 7]
lista[0] = 10
print(lista)
```

```python
# las tuplas son inmutables, no se pueden modificar sus elementos
tupla = (1, 3, 5, 7)
tupla[0] = 10
print(tupla)
```

```python
# count() - Devuelve el número de veces que aparece un elemento en la tupla
tupla.count(5)
```

```python
# index() - Devuelve el índice de la primera aparición de un elemento en la tupla
tupla.index(5)
```

```python
# las tuplas pueden contener elementos de diferentes tipos de datos
tupla = ("maria", 1, 2, 0.555, True, [1, 2, 3], (1, 2, 3))
print(tupla)
```

```python
# Convertir una tupla en una lista
tupla = (1, 3, 5, 7)
lista = list(tupla)
print(lista)
```

```python
# Método para modificar una tupla: convertirla en lista, modificar la lista y luego volver a convertirla en tupla
tupla = (1, 3, 5, 7)
lista = list(tupla)

lista[0] = 10
new_tupla = tuple(lista)
print(new_tupla)
```

```python
# Slicing de tuplas
tupla[1:3]
```
