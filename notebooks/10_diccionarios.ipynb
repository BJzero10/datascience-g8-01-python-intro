### Diccionarios

```python
# estructura de diccionarios
mi_diccionario = {
    "peru": "sol",
    "usa": "dolar",
    "japon": "yen",
    "canada": "dolar canadiense"
}

mi_diccionario
```

```python
# añadir un elemento al diccionario
mi_diccionario["mexico"] = "peso"

mi_diccionario
```

```python
# modificar un elemento del diccionario
mi_diccionario["peru"] = "nuevo sol"
mi_diccionario
```

```python
# metodos de diccionarios
# keys(): Devuelve una vista de las claves del diccionario.
claves = mi_diccionario.keys()
print("Claves del diccionario:", claves)
```

```python
# items(): Devuelve una vista de los pares clave-valor del diccionario.
items = mi_diccionario.items()
print("Pares clave-valor del diccionario:", items)
```

```python
# values(): Devuelve una vista de los valores del diccionario.
valores = mi_diccionario.values()
print("Valores del diccionario:", valores)
```

```python
# get(): Devuelve el valor asociado a una clave específica. Si la clave no existe, devuelve un valor predeterminado (None si no se especifica).
valor_peru = mi_diccionario.get("peru")
print("Valor asociado a la clave 'peru':", valor_peru)
print("Valor asociado a la clave 'brasil' (no existe):", mi_diccionario.get("brasil"))
print("Valor asociado a la clave 'brasil' (no existe, con valor predeterminado):", mi_diccionario.get("brasil", "No disponible"))
```

```python
# update(): Actualiza el diccionario con los pares clave-valor de otro diccionario.
nuevo_diccionario = {
    "brasil": "real",
    "argentina": "peso argentino"
}
mi_diccionario.update(nuevo_diccionario)
mi_diccionario
```

```python
# setdefault(): Devuelve el valor asociado a una clave específica. Si la clave no existe, agrega la clave con un valor predeterminado.
valor_chile = mi_diccionario.setdefault("chile", "peso chileno")
print("Valor asociado a la clave 'chile':", valor_chile)
print("Diccionario después de usar setdefault:", mi_diccionario)
mi_diccionario
```

```python
# pop(): Elimina un elemento del diccionario y devuelve su valor. Si la clave no existe, se genera un error.
valor_eliminado = mi_diccionario.pop("japon")
print("Valor eliminado asociado a la clave 'japon':", valor_eliminado)
mi_diccionario
```

```python
# len(): Devuelve el número de elementos en el diccionario.
tamaño = len(mi_diccionario)
print("Número de elementos en el diccionario:", tamaño)
```

```python
# recorrido de diccionarios
for clave, valor in mi_diccionario.items():
    print(f"Clave: {clave}, Valor: {valor}")
```
