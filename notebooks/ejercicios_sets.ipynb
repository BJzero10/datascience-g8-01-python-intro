#### 1. Estudiantes inscritos en dos cursos
estudiantes_python = { "Ana", "Luis", "María", "Carlos", "Sofía", "Diego" }

estudiantes_javascript = { "María", "Carlos", "Pedro", "Lucía", "Diego", "Valeria" }

Encuentra:
- Los estudiantes inscritos en ambos cursos.
- Los estudiantes que solo están en Python.
- Los estudiantes que solo están en JavaScript.
- Todos los estudiantes sin repetir nombres.

```python
estudiantes_python = { "Ana", "Luis", "María", "Carlos", "Sofía", "Diego" }
estudiantes_javascript = { "María", "Carlos", "Pedro", "Lucía", "Diego", "Valeria" }

interseccion_estudiantes = estudiantes_python.intersection(estudiantes_javascript)
print("Estudiantes que estudian tanto Python como JavaScript:", interseccion_estudiantes)
solo_python = estudiantes_python.difference(estudiantes_javascript)
print("Estudiantes que solo estudian Python:", solo_python)
solo_javascript = estudiantes_javascript.difference(estudiantes_python)
print("Estudiantes que solo estudian JavaScript:", solo_javascript)
union_estudiantes = estudiantes_python.union(estudiantes_javascript)
print("Todos los estudiantes (Python o JavaScript):", union_estudiantes)
```

#### 2. Comparar inventarios de dos almacenes
almacen_arequipa = { "laptop", "teclado", "mouse", "monitor", "impresora", "audifonos" }

almacen_lima = { "laptop", "mouse", "monitor", "tablet", "webcam", "parlantes" }

Determina:
- Los productos disponibles en ambos almacenes. 
- Los productos exclusivos del almacén de Arequipa.
- Los productos exclusivos del almacén de Lima.
- Todos los productos disponibles sin repetir.
- Si ambos almacenes tienen exactamente el mismo inventario.

```python
almacen_arequipa = { "laptop", "teclado", "mouse", "monitor", "impresora", "audifonos" }
almacen_lima = { "laptop", "mouse", "monitor", "tablet", "webcam", "parlantes" }

todos_productos = almacen_arequipa.union(almacen_lima)
print("Todos los productos en Arequipa y Lima:", todos_productos)
solo_arequipa = almacen_arequipa.difference(almacen_lima)
print("Productos solo en Arequipa:", solo_arequipa)
solo_lima = almacen_lima.difference(almacen_arequipa)
print("Productos solo en Lima:", solo_lima)
print("Todos los productos disponibles sin repetir:", todos_productos)
mismo_inventario = almacen_arequipa == almacen_lima
print(f"¿Los almacenes de Arequipa y Lima tienen el mismo inventario?: {mismo_inventario}")
```

#### 3. Validar permisos de un usuario

permisos_usuario = { "leer", "editar", "comentar", "descargar" }

permisos_requeridos = { "leer", "editar", "eliminar", "publicar" }

Crea un programa que muestre:
- Si el usuario posee todos los permisos requeridos.
- Los permisos que le faltan.
- Los permisos adicionales que tiene.
- La cantidad total de permisos diferentes.

```python
permisos_usuario = { "leer", "editar", "comentar", "descargar" }
permisos_requeridos = { "leer", "editar", "eliminar", "publicar" }

posee_todos_requeridos = permisos_requeridos.issubset(permisos_usuario)
print(f"¿El usuario posee todos los permisos requeridos?: {posee_todos_requeridos}")

permisos_faltantes = permisos_requeridos - permisos_usuario
print("Permisos requeridos que el usuario no posee:", permisos_faltantes)

permisos_adicionales = permisos_usuario - permisos_requeridos
print("Permisos adicionales que el usuario posee:", permisos_adicionales)

permisos_diferentes = permisos_usuario.symmetric_difference(permisos_requeridos)
print("Permisos que son diferentes entre el usuario y los requeridos:", permisos_diferentes)
```

#### 4. Números presentes en varias listas

numeros_a = {2, 4, 6, 8, 10, 12, 14}

numeros_b = {4, 6, 9, 10, 13, 14, 16}

numeros_c = {1, 4, 6, 10, 11, 14, 18}

Encuentra:
- Los números presentes en los tres conjuntos.
- Los números presentes en numeros_a y numeros_b, pero no en numeros_c.
- Los números que solo pertenecen a numeros_a.
- Todos los números diferentes.

```python
numeros_a = {2, 4, 6, 8, 10, 12, 14}
numeros_b = {4, 6, 9, 10, 13, 14, 16}
numeros_c = {1, 4, 6, 10, 11, 14, 18}

numeros_interseccion = numeros_a.intersection(numeros_b, numeros_c)
print("Números comunes en los tres conjuntos:", numeros_interseccion)

interseccion_ab_menos_c = numeros_a.intersection(numeros_b).difference(numeros_c)
print("Números comunes en A y B pero no en C:", interseccion_ab_menos_c)

solo_a = numeros_a.difference(numeros_b, numeros_c)
print("Números que están solo en A:", solo_a)

solo_diferentes = numeros_a.symmetric_difference(numeros_b).symmetric_difference(numeros_c)
print("Números que son diferentes entre los tres conjuntos:", solo_diferentes)
```

```python
print("interseccion A, B y C:", numeros_a & numeros_b & numeros_c)
print("interseccion A y B menos C:", (numeros_a & numeros_b) - numeros_c)
print("solo A:", numeros_a - numeros_b - numeros_c)
print("solo diferentes:", numeros_a ^ numeros_b ^ numeros_c)
print("union A, B y C:", numeros_a | numeros_b | numeros_c)
```

#### 5. Comparar habilidades de candidatos

habilidades_candidato_1 = { "Python", "Git", "SQL", "Docker", "React" }

habilidades_candidato_2 = { "Python", "JavaScript", "React", "Node.js", "Git" }

habilidades_requeridas = { "Python", "Git", "SQL", "Docker" }

Determina:
- Las habilidades que comparten ambos candidatos.
- Las habilidades exclusivas de cada candidato.
- Qué habilidades requeridas le faltan a cada candidato.
- Qué candidato cumple con más requisitos.
- Todas las habilidades disponibles sin repetir.

```python
habilidades_candidato_1 = { "Python", "Git", "SQL", "Docker", "React" }
habilidades_candidato_2 = { "Python", "JavaScript", "React", "Node.js", "Git" }
habilidades_requeridas = { "Python", "Git", "SQL", "Docker" }

interseccion_candidatos = habilidades_candidato_1.intersection(habilidades_candidato_2)
print("Habilidades comunes entre los dos candidatos:", interseccion_candidatos)

exclusivas_candidato_1 = habilidades_candidato_1.difference(habilidades_candidato_2)
print("Habilidades exclusivas del candidato 1:", exclusivas_candidato_1)

exclusivas_candidato_2 = habilidades_candidato_2.difference(habilidades_candidato_1)
print("Habilidades exclusivas del candidato 2:", exclusivas_candidato_2)

falta_candidato_1 = habilidades_requeridas.difference(habilidades_candidato_1)
print("Habilidades requeridas que le faltan al candidato 1:", falta_candidato_1)

falta_candidato_2 = habilidades_requeridas.difference(habilidades_candidato_2)
print("Habilidades requeridas que le faltan al candidato 2:", falta_candidato_2) 
```

```python
if len(falta_candidato_1) < len(falta_candidato_2):
    print("El candidato 1 tiene más habilidades requeridas que el candidato 2.")
elif len(falta_candidato_2) < len(falta_candidato_1):
    print("El candidato 2 tiene más habilidades requeridas que el candidato 1.")
else:
    print("Ambos candidatos tienen la misma cantidad de habilidades requeridas que les faltan.")

union_habilidades = habilidades_candidato_1 | habilidades_candidato_2 | habilidades_requeridas
print("Todas las habilidades disponibles:", union_habilidades)
```
