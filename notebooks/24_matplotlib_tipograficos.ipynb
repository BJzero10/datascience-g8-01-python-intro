### Gráfico de líneas

- Se usa cuando el eje X sigue un orden temporal, natural: dias, meses, distancia, temperatura creciente, iteracion de un algoritmo, etc.

```python
from matplotlib import pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 45]


plt.plot(x, y, marker='o', linestyle='-', color='b')
plt.xlabel('Periodo')
plt.ylabel('Valor')
plt.title('Ejemplo de gráfico de líneas')
plt.show()
```

### Gráfico de barras

- Sirve para comparar magnitudes, y responder preguntas como qué categoría vende más?, qué clase tiene mayor promedio?, etc.
- Compara cantidades

```python
categorias = ['A', 'B', 'C', 'D']
valores = [40, 70, 30, 90]

plt.bar(categorias, valores, width=0.5, label=categorias, color=['red', 'black', 'blue', 'green'])
plt.xlabel('Categoria')
plt.ylabel('Frecuencia')
plt.title('Gráfico de barras verticales')
plt.legend(title='Categorias')
plt.show()
```

```python
plt.barh(categorias, valores, color='green')
plt.xlabel('Frecuencia')
plt.ylabel('Categoria')
plt.title('Gráfico de barras horizontal')
plt.show()
```

### Gráfico pastel o pie chart

```python
frutas = ['Manzanas', 'Plátanos', 'Peras']
porcentajes = [30, 50, 20]

plt.pie(porcentajes, labels=frutas, autopct='%1.1f%%')
plt.title('Gráfico de pastel')
plt.show()
```

### Gráfico de dispersión (scatter plot)

- Este gráfico me permite visualizar la correlación, fuerza y relación entre 2 variables y también puede mostrarme valores atípicos

```python
import numpy as np

x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x, y, color='blue')
plt.xlabel('Variable (x)')
plt.ylabel('Variable (y)')
plt.title('Gráfico de dispersión')
plt.show()
```

### Gráfico Histograma

- Reconocer concentración, dispérsion y asimetría. El número de bins importa, muy pocos bins me quita detalle, pero muchos bins me da ruido visual

```python

datos = np.random.randn(1000)
plt.hist(datos, bins=20, color='orange', edgecolor='black' )
plt.xlabel('Valor')
plt.ylabel('Frecuencia')
plt.title('Histograma')
plt.show()
```

### Gráfico de caja, bigotes o boxplot

```python
data_box = [np.random.normal(0, std, 1000) for std in range(1, 4)]

plt.boxplot(data_box, orientation='vertical', patch_artist=True)
plt.xlabel('Grupo')
plt.ylabel('Valor')
plt.title('Gráfico Boxplot')
plt.show()
```
