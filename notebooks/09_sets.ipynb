```python
# sets no permiten elementos duplicados y no tienen un orden específico.
conjunto_a = {1, 3, 5, 7, 9, 9, 1, 3}
conjunto_b = {3, 5, 7}
conjunto_c = [1, 3, 5, 7, 9, 9, 1, 3]

print("Conjunto A:", conjunto_a)
print("Conjunto B:", conjunto_b)
print("Conjunto C:", conjunto_c)
```

```python
# union de conjuntos
conjunto_a = {1, 3, 5, 7, 9, 9, 1, 3}
conjunto_b = {3, 5, 7}

conjunto_union = conjunto_a.union(conjunto_b)
print("Conjunto A:", conjunto_a)
print("Conjunto B:", conjunto_b)
print("Unión de conjuntos:", conjunto_union)
```

```python
# intersección de conjuntos
conjunto_interseccion = conjunto_a.intersection(conjunto_b)
print("Conjunto A:", conjunto_a)
print("Conjunto B:", conjunto_b)
print("Intersección de conjuntos:", conjunto_interseccion)
```

```python
# diferencia de conjuntos
conjunto_diferencia = conjunto_a.difference(conjunto_b)
print("Conjunto A:", conjunto_a)
print("Conjunto B:", conjunto_b)
print("Diferencia de conjuntos (A - B):", conjunto_diferencia)
```

```python
# diferencia simétrica de conjuntos
conjunto_diferencia_simetrica = conjunto_a.symmetric_difference(conjunto_b)
print("Conjunto A:", conjunto_a)
print("Conjunto B:", conjunto_b)
print("Diferencia simétrica de conjuntos (A Δ B):", conjunto_diferencia_simetrica)
```

```python
# metodos de conjuntos
# isdisjoint(): Devuelve True si dos conjuntos no tienen elementos en común.
conjunto_a = {1, 2, 3}
conjunto_b = {4, 5, 6}
resultado = conjunto_a.isdisjoint(conjunto_b)
print("Conjunto A:", conjunto_a)
print("Conjunto B:", conjunto_b)
print("¿Los conjuntos A y B no tienen elementos en común?", resultado)
```

```python
# add(): Agrega un elemento al conjunto.
conjunto_a.add(4)
print("Conjunto A después de agregar 4:", conjunto_a)
```

```python
# remove(): Elimina un elemento del conjunto. Si el elemento no existe, se genera un error.
conjunto_a.remove(2)
print("Conjunto A después de eliminar 2:", conjunto_a)
```

```python
# issubset(): Devuelve True si todos los elementos del conjunto están en otro conjunto.
conjunto_a = {1, 2, 3}
conjunto_b = {1, 2, 3, 4, 5}
resultado = conjunto_a.issubset(conjunto_b)
print("Conjunto A:", conjunto_a)
print("Conjunto B:", conjunto_b)
print("¿Todos los elementos de A están en B?", resultado)
```

```python
# issuperset(): Devuelve True si todos los elementos de otro conjunto están en el conjunto.
conjunto_a = {1, 2, 3, 4, 5}
conjunto_b = {1, 2, 3}
resultado = conjunto_a.issuperset(conjunto_b)
print("Conjunto A:", conjunto_a)
print("Conjunto B:", conjunto_b)
print("¿Todos los elementos de B están en A?", resultado)
```
