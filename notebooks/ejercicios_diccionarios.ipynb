#### 1. Consolidación de inventarios
Combina tres inventarios provenientes de distintas sucursales. Cuando un producto aparezca en varias sucursales, suma sus cantidades y calcula también en cuántas sucursales está disponible.

inventario_norte = {"laptop": 4, "mouse": 10, "monitor": 3}

inventario_sur = {"laptop": 2, "teclado": 8, "monitor": 5}

inventario_centro = {"mouse": 6, "teclado": 4, "webcam": 7}

Requisitos:

- Crea un diccionario final con las claves cantidad_total y sucursales.
- No modifiques los diccionarios originales.
- Ordena el resultado alfabéticamente por producto.

```python
inventario_norte = {"laptop": 4, "mouse": 10, "monitor": 3}
inventario_sur = {"laptop": 2, "teclado": 8, "monitor": 5}
inventario_centro = {"mouse": 6, "teclado": 4, "webcam": 7}

# combinar los inventarios en un solo diccionario
inventario_total = {}
for inventario in [inventario_norte, inventario_sur, inventario_centro]:
    for item, cantidad in inventario.items():
        if item in inventario_total:
            inventario_total[item] += cantidad
        else:
            inventario_total[item] = cantidad

# inventario total ordenado
diccionario_ordenado = dict(
    sorted(inventario_total.items(), key= lambda item: item[0])
)
diccionario_ordenado
```

#### 2. Ranking de estudiantes con desempate
Calcula el promedio de cada estudiante y genera un ranking de mayor a menor. Si dos estudiantes tienen el mismo promedio, ordénalos alfabéticamente.

calificaciones = { "Ana": [17, 18, 16, 19], "Luis": [15, 16, 17, 16], "Carla": [18, 17, 19, 18], "Pedro": [16, 16, 16, 16], "Lucía": [15, 17, 16, 16] }

Requisitos:

- Guarda promedio, nota_máxima y nota_mínima por estudiante.
- El ranking final debe ser una lista de tuplas.

```python
calificaciones = {
    "Ana": [17, 18, 16, 19],
    "Luis": [15, 16, 17, 16],
    "Carla": [18, 17, 19, 18],
    "Pedro": [16, 16, 16, 16],
    "Lucía": [15, 17, 16, 16]
    }

ranking_estudiantes = {}
for estudiante, notas in calificaciones.items():
    # guardamos promedio
    promedio = sum(notas) / len(notas)
    # guardamos nota_maxima
    nota_max = max(notas)
    # guardamos nota_minima
    nota_min = min(notas)
    ranking_estudiantes[estudiante] = [promedio, nota_max, nota_min]

ranking_final = sorted(ranking_estudiantes.items(), key=lambda item: (-item[1][0], item[0]))
ranking_final
```

#### 4. Análisis de frecuencia de palabras
Construye un diccionario de frecuencias ignorando mayúsculas, signos de puntuación y palabras vacías.

texto = "Python es poderoso; Python es flexible, y aprender Python puede ser muy divertido." 

palabras_vacias = {"es", "y", "ser", "muy", "puede"}

Requisitos:

- Muestra las tres palabras más frecuentes.
- Resuelve el ejercicio sin usar collections.Counter.

```python
texto = "Python es poderoso; Python es flexible, y aprender Python puede ser muy divertido." 
palabras_vacias = {"es", "y", "ser", "muy", "puede"}

diccionario_frecuencias = {}

texto_limpio = texto.replace(";","").replace(",","").replace(".","")

for word in texto_limpio.lower().split():
    if word not in palabras_vacias:
        if word in diccionario_frecuencias.keys():
            diccionario_frecuencias[word] += 1
        else:
            diccionario_frecuencias[word] = 1

diccionario_frecuencias

diccionario_ordenado = sorted(diccionario_frecuencias.items(), key=lambda item: (-item[1], item[0]))

print(f"Las 3 palabras más frecuentes son: {diccionario_ordenado[:3]}")
```

#### 7. Sistema de votación con validación
Procesa una lista de votos y separa los votos válidos de los inválidos.

candidatos = {"Ana", "Luis", "Carla"}

votos = ["Ana", "Luis", "ana", "Carla", "Pedro", "Luis", "ANA", "", "Carla"]

Requisitos:

- Los votos deben ser insensibles a mayúsculas.
- Registra conteo por candidato, votos inválidos y porcentaje obtenido.
- Determina si hay ganador o empate.

```python
candidatos = {"Ana", "Luis", "Carla"}
votos = ["Ana", "Luis", "ana", "Carla", "Pedro", "Luis", "ANA", "", "Carla"]

resumen_votos = {}

for voto in votos:
    voto_limpio = voto.strip().capitalize()

    if voto_limpio in candidatos:
        resumen_votos.setdefault(voto_limpio, 0)
        resumen_votos.setdefault("pct_ganador", 0)
        resumen_votos.setdefault("votos_invalidos", 0)
        resumen_votos[voto_limpio] += 1
    else:
        resumen_votos["votos_invalidos"] += 1

ganador = max(resumen_votos.items(), key= lambda item:item[1])
print(ganador)
resumen_votos["pct_ganador"] = round(ganador[1] * 100 / len(votos), 2)
print(resumen_votos)
```

#### 9. Control de stock con transacciones
Aplica una secuencia de movimientos de entrada y salida sobre un inventario.

stock = {"laptop": 5, "mouse": 12, "monitor": 4}
movimientos = [
    {"producto": "laptop", "tipo": "salida", "cantidad": 2},
    {"producto": "mouse", "tipo": "entrada", "cantidad": 5},
    {"producto": "monitor", "tipo": "salida", "cantidad": 6}, {"producto": "teclado", "tipo": "entrada", "cantidad": 3}
]

Requisitos:

- No permitas que el stock quede negativo.
- Registra los movimientos rechazados en una lista.
- Crea productos nuevos solo mediante entradas.

```python
stock = {"laptop": 5, "mouse": 12, "monitor": 4}
movimientos = [
    {"producto": "laptop", "tipo": "salida", "cantidad": 2},
    {"producto": "mouse", "tipo": "entrada", "cantidad": 5},
    {"producto": "monitor", "tipo": "salida", "cantidad": 6}, {"producto": "teclado", "tipo": "entrada", "cantidad": 3}
]

movimientos_rechazados = []

for m in movimientos:
    prod = m.get("producto")
    tipo = m.get("tipo")
    cant = m.get("cantidad", 0)

    if tipo == "entrada":
        stock[prod] = stock.get(prod, 0) + cant
    elif tipo == "salida":
        if prod not in stock:
            movimientos_rechazados.append({"movimiento": m, "motivo": "producto no existe"})
        elif stock[prod] < cant:
            movimientos_rechazados.append({"movimiento": m, "motivo": "stock insuficiente"})
        else:
            stock[prod] -= cant
    else:
        movimientos_rechazados.append({"movimiento": m, "motivo": "tipo desconocido"})

{"stock_final": stock, "movimientos_rechazados": movimientos_rechazados}
```

#### 10. Matriz dispersa con diccionarios
Representa y manipula una matriz dispersa utilizando un diccionario cuyas claves sean tuplas (fila, columna).

matriz = {
    (0, 1): 5,
    (1, 3): 8,
    (2, 0): 4,
    (3, 2): 7
}

filas = 4

columnas = 4

Requisitos:

- Convierte la matriz dispersa en una lista de listas.
- Calcula la suma por fila y por columna.
- Obtén la matriz transpuesta como otro diccionario.

```python
matriz = {
    (0, 1): 5,
    (1, 3): 8,
    (2, 0): 4,
    (3, 2): 7
}
filas = 4
columnas = 4

matriz_densa = [[0 for _ in range(columnas)] for _ in range(filas)]
for (r, c), val in matriz.items():
    matriz_densa[r][c] = val

suma_filas = [sum(fila) for fila in matriz_densa]
suma_columnas = [sum(matriz_densa[r][c] for r in range(filas)) for c in range(columnas)]

matriz_transpuesta = {(c, r): val for (r, c), val in matriz.items()}

{"matriz_densa": matriz_densa, "suma_filas": suma_filas, "suma_columnas": suma_columnas, "matriz_transpuesta": matriz_transpuesta}
```

#### 11. Historial de precios
Analiza el historial de precios de varios productos.

historial = {
    "laptop": [3500, 3450, 3600, 3400],
    "mouse": [80, 85, 78, 82],
    "monitor": [900, 920, 910, 880]
}

Requisitos:

- Calcula precio mínimo, máximo, promedio y variación porcentual entre el primer y último valor.
- Identifica el producto con mayor reducción porcentual.

```python
historial = {
    "laptop": [3500, 3450, 3600, 3400],
    "mouse": [80, 85, 78, 82],
    "monitor": [900, 920, 910, 880]
}

estadisticas_precios = {}
for producto, precios in historial.items():
    primero, ultimo = precios[0], precios[-1]
    minimo = min(precios)
    maximo = max(precios)
    promedio = round(sum(precios) / len(precios), 2)
    variacion_pct = round((ultimo - primero) / primero * 100, 2) if primero != 0 else None
    estadisticas_precios[producto] = {
        "minimo": minimo,
        "maximo": maximo,
        "promedio": promedio,
        "variacion_pct": variacion_pct
    }

# producto con mayor reducción porcentual (más negativo)
reducciones = [(p, d["variacion_pct"]) for p, d in estadisticas_precios.items() if d["variacion_pct"] is not None]
producto_mayor_reduccion = min(reducciones, key=lambda x: x[1]) if reducciones else (None, None)

{"estadisticas": estadisticas_precios, "mayor_reduccion": {"producto": producto_mayor_reduccion[0], "reduccion_pct": producto_mayor_reduccion[1]}}
```

#### 18. Contar frutas repetidas
